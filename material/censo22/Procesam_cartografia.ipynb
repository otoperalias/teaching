## Procesamiento de la Cartografía del Censo 2022 y la extracción de los sectores de la provincia de Loja.

#### 1. Importamos las librerías que vamos a usar

````python
import pandas as pd
import geopandas as gpd
````

#### 2. Descarga y lectura de datos

* En primer lugar, se descarga la carpeta comprimida con la capa de sectores del censo anonimizados: https://www.ecuadorencifras.gob.ec/documentos/web-inec/capa/CapaSectores.zip y se sube a Golab *sin descomprimir*, en la pestaña de archivos. Como es un archivo de gran tamaño, tomará varios minutos.
* A continuación, se lee usando geopandas y la función ```read_file()```. Debido al tamaño del archivo, también tomará algún tiempo (~5 minutos).

````python
sectores=gpd.read_file('zip://CapaSectores.zip!sectores_anonimizados.gpkg', layer="sectores_anonimizados_indicadores_censo_2022")
````

#### 3. Seleccionamos la provincia de Loja


````python
sect_loja=sectores.loc[sectores['provincia']=='11'].copy()
````

````python
#Echamos un vistazo
sect_loja
````

#### 4. Guardamos la capa
Guardamos la capa en formato .shp como archivo comprimido.


````python
sect_loja.to_file("sect_loja.shp.zip")
````

Finalmente, lo descargamos en nuestro ordenador.
