### Este notebook importa la tabla de viviendas del censo de 2022 y selecciona la provincia de Loja

#### 1. Descarga de microdatos del Censo de 2022

* En este enlace https://www.censoecuador.gob.ec/data-censo-ecuador/, clicamos en el primer icono de "Base de datos" y luego en "Base de datos CSV" en 2022.  
* Se descarga una carpeta comprimida de 474 Megas. Descomprimimos esta carpeta, que contiene 5 archivos con formato .csv.  
* Subimos a Google Colab el archivo "CPV_Vivienda_2022_Nacional", que pesa 647 megas y por tanto tarda algunos minutos en subirse.

#### 2. Importamos las librerías que vamos a usar en este notebook


````python
import pandas as pd
````

#### 3. Leemos con ```pandas``` el fichero y lo exploramos


````python
# Leemos el fichero
viv=pd.read_csv("CPV_Vivienda_2022_Nacional.csv", sep=";")
````

````python
# Mostramos la tabla
viv
````

````python
# Vemos que tiene 6.6 millones de registros (!!)
````

````python
# Listamos las columnas de la tabla
viv.columns
````

#### 4. Seleccionamos los registros de Loja


* El archivo "4_DICCIONARIO_DE_VARIABLES_CPV_2022" (disponible [aquí](https://www.censoecuador.gob.ec/data-censo-ecuador/)) indica que, en la tabla de vivienda, la columna I01 hace referencia al código de la provincia.
* El archivo "6.-ANEXO_CLASIFICADORES_NOMENCLATURAS" (disponible también [aquí](https://www.censoecuador.gob.ec/data-censo-ecuador/)) indica que el código de Loja es el 11.



````python
loja=viv.loc[viv['I01']==11].copy()
````

````python
# Echamos un vistazo a la nueva tabla para Loja, que ahora tiene 199000 registros.
loja
````

#### 5. Guardamos la tabla en un archivo csv para descargarlo en nuestro ordenador.

````python
loja.to_csv("viv_loja.csv")
````
